### HTTP

HTTP stands for HyperText Transfer Protocol.
It is the communication protocol that the web uses for sending and receiving information between:

- a client (usually a browser), and
- a server (a computer hosting a website).

How it works (simple idea)

You (the client) request something → e.g., you type a URL.

Your browser sends an HTTP request to a server.

The server processes the request.

The server sends back an HTTP response, usually containing:

a status code (200 OK, 404 Not Found…)

headers (extra info)

a body (HTML, JSON, images, files…)

This exchange happens very fast.

Example

When you visit https://example.com:

Your browser sends:
GET / HTTP/1.1

The server answers:

HTTP/1.1 200 OK
Content-Type: text/html


(and then sends the HTML of the page)

Key ideas
1. Stateless

HTTP doesn’t remember anything between requests.
Each request is independent unless you use cookies or sessions.

2. Methods

These are the actions you request from the server:

GET → retrieve data

POST → send data (forms, login…)

PUT → update data

DELETE → delete data

3. Status codes

Servers answer with these codes:

200 OK → everything worked

301/302 → redirect

404 Not Found → page doesn’t exist

500 Internal Server Error → server crashed



HTTP is the protocol that defines how browsers and servers communicate on the web.
